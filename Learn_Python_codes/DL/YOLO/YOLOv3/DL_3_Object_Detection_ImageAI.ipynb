```python
import os
from imageai.Detection import ObjectDetection

# ۱. راه‌اندازی و بارگذاری مدل
detector = ObjectDetection()
detector.setModelTypeAsYOLOv3()
detector.setModelPath("/content/drive/MyDrive/model_path/yolov3.pt")
detector.loadModel()

# ۲. مشخص کردن مسیر تصاویر
images_path = "/content/drive/MyDrive/images"

# ۳. تنظیم فیلتر سفارشی (مثلاً فقط خودرو)
custom_detector = detector.CustomObjects(car=True)

# ۴. اجرای تشخیص روی تصویر
custom_detection = detector.detectObjectsFromImage(
    custom_objects=custom_detector,
    input_image=os.path.join(images_path, "04_tehran.png"),
    output_image_path=os.path.join(images_path, "04_tehran_customdetected.jpg"),
    minimum_percentage_probability=80
)

# ۵. چاپ نتایج
for obj in custom_detection:
    print(obj["name"], " : ", obj["percentage_probability"], "%")
```

```python
import os
import cv2
from imageai.Detection import VideoObjectDetection

video_detector.setModelPath(os.path.join(model_path, "yolov3.pt"))
video_detector.loadModel()

camera = cv2.VideoCapture(0)

vid_detection = video_detector.detectObjectsFromVideo(
    camera_input=camera,
    output_file_path=os.path.join(video_path, "camera_detected_video"),
    frames_per_second=20,
    log_progress=True
)

print(vid_detection)
camera.release()
```
